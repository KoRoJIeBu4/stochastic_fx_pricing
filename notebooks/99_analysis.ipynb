```python
from stochastic_fx_pricing.config import load_models
from stochastic_fx_pricing.simulation import SDESimulation

models = load_models()

range_accrual_contract = models["range_accrual_contract"]
simulation_config = models["simulation_config"]
market_model_config = models["market_model_config"]

simulation = SDESimulation(
    range_accrual_contract=range_accrual_contract,
    simulation_config=simulation_config,
    market_model_config=market_model_config,
)

trajectories = simulation.run()
```

```python
from stochastic_fx_pricing.plots import plot_trajectories

plot_trajectories(
    trajectories=trajectories,
    max_trajectories=simulation_config.max_plot_trajectories,
    seed=simulation_config.seed,
)
```
